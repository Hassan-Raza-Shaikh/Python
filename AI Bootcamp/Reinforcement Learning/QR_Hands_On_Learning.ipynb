**Task 1.2 — Q-Learning Update Equation**

```python
import numpy as np

Q = np.zeros((5,4))

state = 1
action = 2
next_state = 3

reward = 10

alpha = 0.1
gamma = 0.9

# Current value
current = Q[state, action]

# Best future value
future = np.max(Q[next_state])

# Complete the update

Q[state, action] = current + alpha * (
    reward + gamma * future - current
)

print(Q)
```

```python
import numpy as np
import matplotlib.pyplot as plt



GRID_SIZE = 5

START = (0,0)

GOAL = (4,4)

HOLES = {
    (1,3),
    (2,2),
    (3,1)
}

actions = [
    (-1,0),   # Up
    (1,0),    # Down
    (0,-1),   # Left
    (0,1)     # Right
]

alpha = 0.1
gamma = 0.95
epsilon = 0.2

Q = np.zeros((GRID_SIZE,GRID_SIZE,4))

def step(state,action):

    r,c = state

    dr,dc = actions[action]

    nr = max(0,min(GRID_SIZE-1,r+dr))
    nc = max(0,min(GRID_SIZE-1,c+dc))

    new_state = (nr,nc)

    if new_state in HOLES:
        return new_state,-10,True

    if new_state==GOAL:
        return new_state,20,True

    return new_state,-1,False


rewards=[]

for episode in range(500):

    state = START

    total_reward = 0

    done=False

    while not done:

        if np.random.rand()<epsilon:
            action=np.random.randint(4)
        else:
            action=np.argmax(Q[state])

        next_state,reward,done=step(state,action)

        best=np.max(Q[next_state])

        Q[state][action]+=alpha*(
            reward+gamma*best-Q[state][action]
        )

        state=next_state

        total_reward+=reward

    rewards.append(total_reward)

# Plot Training Reward

plt.figure(figsize=(8,4))
plt.plot(rewards)
plt.xlabel("Episode")
plt.ylabel("Reward")
plt.title("Training Rewards")
plt.grid(True)
plt.show()

# Test Learned Policy

state = START

path=[state]

done=False

while not done:

    action=np.argmax(Q[state])

    state,reward,done=step(state,action)

    path.append(state)

print("\nLearned Path:")
print(path)
```
