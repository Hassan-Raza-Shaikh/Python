# Model Building and Hyperparameter tunning

```python
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 16),
            nn.ReLU(),
            nn.Linear(16, 2)
        )

    def forward(self, x):
        return self.net(x)
```

```python
def train(model, optimizer, epochs=50):
    criterion = nn.CrossEntropyLoss()
    losses = []

    for epoch in range(epochs):
        epoch_loss = 0

        for xb, yb in loader:
            optimizer.zero_grad()        # clear old gradients
            preds = model(xb)            # forward
            loss = criterion(preds, yb)  # loss
            loss.backward()              # gradients
            optimizer.step()             # update weights

            epoch_loss += loss.item()

        losses.append(epoch_loss / len(loader))

    return losses
```

Technique 1: Constant Learning Rate (Baseline)
Logic

Same η throughout training

Simple but inefficient

```python
model = SimpleNN()
optimizer = optim.SGD(model.parameters(), momentum=0.9, lr=0.1)#momentum=0.9,

losses_constant = train(model, optimizer)
losses_constant
```

```python
plt.plot(losses_constant)
plt.title("Loss vs Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

Problem

Cannot adapt to loss surface

Often overshoots minima

Technique 3: Step Decay Learning Rate
Idea

Reduce LR after fixed number of epochs

$$
η
0
=0.1
$$
$$ s=10 $$ $$ \gamma=0.1 $$

Then:

| Epoch | Learning Rate |
| ----: | ------------: |
|   0–9 |       \(0.1\) |
| 10–19 |      \(0.01\) |
| 20–29 |     \(0.001\) |
| 30–39 |    \(0.0001\) |

Epoch 5
$$ \eta_5=0.1(0.1)^{\lfloor5/10\rfloor} $$ $$ =0.1(0.1)^0 $$ $$ \boxed{0.1} $$
Epoch 10
$$ \eta_{10}=0.1(0.1)^{\lfloor10/10\rfloor} $$ $$ =0.1(0.1)^1 $$ $$ \boxed{0.01} $$
Epoch 20
$$ \eta_{20}=0.1(0.1)^{\lfloor20/10\rfloor} $$ $$ =0.1(0.1)^2 = \boxed{0.001} $$
	​


```python
model = SimpleNN()
optimizer = optim.SGD(model.parameters(), lr=0.1)

scheduler = optim.lr_scheduler.StepLR(
    optimizer,
    step_size=20,
    gamma=0.5
)

criterion = nn.CrossEntropyLoss()

for epoch in range(50):
    for xb, yb in loader:
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)
        loss.backward()
        optimizer.step()

    scheduler.step()
```

Technique 6: Cyclic Learning Rate (Advanced)
Idea

LR oscillates instead of monotonically decreasing

```python
optimizer = optim.SGD(model.parameters(), lr=0.01)

scheduler = optim.lr_scheduler.CyclicLR(
    optimizer,
    base_lr=0.001,
    max_lr=0.1,
    step_size_up=10, #10 batches
    mode='triangular2'
)
```

| Mode        | Description        |
| ----------- | ------------------ |
| triangular  | Linear up & down   |
| triangular2 | Decaying amplitude |
| exp_range   | Exponential decay  |


```python
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=100,
    eta_min=1e-5
)
```

# Learning rate controls convergence behavior

# Same model + same data → LR alone can change results drastically

```python
optimizer = optim.SGD(model.parameters(), lr=0.1)
optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9)
optimizer = optim.RMSprop(model.parameters(), lr=0.001)
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

| Optimizer          | Main idea                                     | Memory                | Main advantage               | Main weakness                  |
| ------------------ | --------------------------------------------- | --------------------- | ---------------------------- | ------------------------------ |
| **SGD**            | Use current gradient                          | None                  | Simple, stable               | Can be slow/zig-zag            |
| **SGD + Momentum** | Add previous update direction                 | Direction             | Faster, reduces zig-zag      | Needs tuning                   |
| **AdaGrad**        | Adapt LR using **all past squared gradients** | All history           | Good for sparse features     | LR can become too small        |
| **RMSProp**        | Adapt LR using **recent squared gradients**   | Recent magnitude      | Fixes AdaGrad's shrinking LR | No explicit momentum direction |
| **Adam**           | **Momentum + RMSProp**                        | Direction + magnitude | Fast, adaptive, widely used  | More hyperparameters           |
